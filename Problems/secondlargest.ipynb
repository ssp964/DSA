## Get the second largest element form the list

```python
import random

list = [ random.randint(1, 100) for i in range(20000000) ]
# list = [20, 20, 20]
#list
```

```python
#Brute Force time complexity is n^2
def find_max(ip_ls):
    '''
    gets the maximum number form the list
    '''
    res = ip_ls[0]
    for i in range(1, len(ip_ls)):
        res = max(res, ip_ls[i])
    return res

def find_sec_max(ip_ls):
    if len(ip_ls) <= 1:
        return None
    max_no = find_max(ip_ls)
    res = None
    for i in ip_ls: 
        if i != max_no: 
            if res == None:
                res = i
            else:
                res = max(res, i)
    return res

print(f"The largest number is : {find_max(list)}")
print(f"The seconfd largest number is : {find_sec_max(list)}")
```

```python
# efficient solution (one traversal)
# 1. chekc if the list has only one element, return None
# 2. iterate through the list
# 3. find the max_no smaller than the current number x, then x will get stored to max_no and max_no will be stored to res
# 4. If the number is not equal to max_no, update res if it's None or if i is greater than res

def getSecMax(ip_list):
    if len(ip_list) <= 1:
        return None
    max_no = ip_list[0]
    res = None
    for i in ip_list[1:]:
        if i > max_no: # if it finds a larger number x then x will get stored to max_no and max_no will be stored to res
            res = max_no
            max_no = i
        elif i != max_no: # If the number is not equal to max_no, update res if it's None or if i is greater than res
            if res == None or res < i:
                res = i
    return res

print(f"Teh second largest number is : {getSecMax(list)}") 
```
