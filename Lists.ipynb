```python
# Reference Assignment: When you assign check2 = check1, both check1 and check2 refer to the same list object in memory. 
# Therefore, any changes made to check2 will also reflect in check1 because they are both pointing to the same list.

# Slicing: When you create check3 with check3 = check1[:], you are creating a new list that is a shallow copy of check1. 
# This means that check3 is a separate list in memory, and changes to check3 will not affect check1 or check2.

check1 = ['Learn', 'Quiz', 'Practice', 'Contribute'] 
check2 = check1 
check3 = check1[:] 

print(check1)
print(check2)
print(check3)

check2[0] = 'Code'
check3[1] = 'Mcq'
print("\n",check1)
print(check2)
print(check3)
count = 0
for c in (check1, check2, check3): 
	if c[0] == 'Code': 
		count += 1
	if c[1] == 'Mcq': 
		count += 10

print (count)
```
